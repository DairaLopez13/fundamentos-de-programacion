```python
def sumar():
    total = 0
    for n in num:
        total += n
    return total

def tuplas():
    num = (2, 4, 5, 7, 13, 3, 1)
    print("Tuplas")
    print(f"La tupla original: {num}")
    print(f"El tercer elemento es: {num[2]}")

    num1 = float(input("Ingrese el primer numero para agregar: "))
    num2 = float(input("Ingrese el segundo numero para agregar: "))
    nueva_tupla = num + (num1, num2)
    print(f"La nueva tupla: {nueva_tupla}")
    lista_numeros = list(nueva_tupla)
    lista_numeros.sort()
    print(f"La lista ordenada: {lista_numeros}")
    suma = sumar(nueva_tupla)
    print(f"La suma de los numeros es: {suma}")

def buscar_telefono(contactos, nombre):
    return contactos.get(nombre)

def diccionarios():
    contactos = {
        "Papá": "5569058758",
        "Mamá": "8721497641",
        "Pachis": "8265061382",
        "Emily": "5599058450",
        "Frank": "4787296401"
    }
    print("Diccionario")
    usuario_nuevo = input("Ingrese el nombre del nuevo usuario: ")
    telefono_nuevo = input("Ingrese el telefono del usuario: ")
    contactos[usuario_nuevo] = telefono_nuevo
    print("Nombres registrados en el sistema")
    for nombre in contactos.keys():
        print(nombre)
    buscar_usuario = input("Ingrese el nombre de usuario que desea buscar: ")
    telefono = buscar_telefono(contactos, buscar_usuario)
    if telefono:
        print(f"Teléfono: {telefono}")
    else:
        print(f"{buscar_usuario} no está en tus contactos")

def dividir_numeros():
    try:
        numero1 = int(input("Ingrese el primer número entero: "))
        numero2 = int(input("Ingrese el segundo número entero: "))
        print("Suma:", numero1 + numero2)
        resultado = numero1 / numero2
        print("División:", resultado)
    except ValueError:
        print("No es posible realizar tu división debido a que ingresaste un"
        "caracter no válido")
    except ZeroDivisionError:
        print("Porfavor verifica tus datos ya que la división con un número"
        "0 no es posible")

def contar_palabras(mensaje):
    palabras = mensaje.split()
    return len(palabras)

def trabajar_frase():
    mensaje = input("Ingrese el mensaje deseado con un mínimo de 4 palabras: ")
    while len(mensaje.split()) < 4:
        print("Error en la ejecución debido a que el mensaje debe de contener"
        "más de 4 palabras")
        mensaje = input("Ingrese el mensaje deseado con un mínimo de 4"
        "palabras: ")

    print("Longitud:", len(mensaje))
    print("El mensaje en mayúsculas:", mensaje.upper())

    buscar_palabra = input("¿Cúal es la palabra que desea reemplazar?: ")
    nueva_palabra = input("¿Por cúal palabra la desea reemplazar?: ")

    mensaje_modificado = mensaje.replace(buscar_palabra, nueva_palabra)
    print(f"El nuevo mensaje es: {mensaje_modificado}")
    cantidad = contar_palabras(mensaje_modificado)
    print(f"Cantidad de palabras del mensaje es: {cantidad}")

opcion = 0

while opcion != 5:
    print("MENÚ PRINCIPAL")
    print("1. Tuplas")
    print("2. Diccionarios")
    print("3. Excepciones")
    print("4. Frases")
    print("5. Finalizar")
    try:
        opcion = int(input("Seleccione una opción: "))
        if opcion == 1:
            tuplas()
        elif opcion == 2:
            diccionarios()
        elif opcion == 3:
            dividir_numeros()
        elif opcion == 4:
            trabajar_frase()
        elif opcion == 5:
            print("El programa termina")
        else:
            print("Opción inválida")
    except ValueError:
        print("Opción inválida")
```
