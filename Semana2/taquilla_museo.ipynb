```python
PRECIO_GRATIS = 0
PRECIO_MENOR = 30
PRECIO_ADULTO = 45
DESCUENTO_ADULTO_MAYOR = 0.12
DESCUENTO_ESTUDIANTE = 0.10
DESCUENTO_PROFESOR = 0.10
DESCUENTO_NINGUNO = 0.00

contador = 0
total_descuento = 0
total_pagar = 0


print("Bienvenidos al museo de Antropologia e historia")

while True:
  edad = (input("Ingresa la edad del visitante para iniciar "
                  "o fin para cerrar la taquilla: "))
  if edad == "fin":
    break
  edad= int(edad)

  if edad < 0 or edad > 120:
    print("Edad invalida, porfavor verifiqque su respuesta")
    continue

  if edad < 3:
    precio_base = PRECIO_GRATIS
    tipo = "Gratis"
  elif edad < 18:
    precio_base = PRECIO_MENOR
    tipo = "Menor"
  else:
    precio_base = PRECIO_ADULTO
    tipo = "Adulto"


  descuento = 0


  if edad >= 18:
    if edad >= 60:
      tipo = "adulto mayor"
      descuento = DESCUENTO_ADULTO_MAYOR
    else:
      print("1.Estudiante")
      print("2.Profesor")
      print("3.Ninguno")
      opcion = input("Elige la opción que te representa: ")

      if opcion == "1":
        tipo = "Estudiante"
        descuento = DESCUENTO_ESTUDIANTE
      elif opcion == "2":
        tipo = "Profesor"
        descuento = DESCUENTO_PROFESOR
      else:
        tipo = "Ninguno"
        descuento = DESCUENTO_NINGUNO

  monto_descuento = precio_base * descuento
  monto_pagar = precio_base - monto_descuento

  contador += 1
  total_descuento += monto_descuento
  total_pagar += monto_pagar


  print(f"Visitante: {contador} | Edad: {edad} | Tipo: {tipo} |")
  print(f"Base: ${precio_base:.2f} | Descuento: ${monto_descuento:.2f} |")
  print(f"Paga: ${monto_pagar:.2f}")
  print("Gracias por tu visita, disfruta tu estadia")
  print("Siguiente en la fila")

print("Corte de caja")
print(f"Total de visitantes cobrados: {contador}")
print(f"Total de descuento: ${total_descuento:.2f}")
print(f"Monto total de ventas: ${total_pagar:.2f}")
```
