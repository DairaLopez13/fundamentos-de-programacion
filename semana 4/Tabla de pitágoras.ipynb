```python
#Tabla de pitagoras
n = 11
matriz = []


def llenar_matriz(matriz):
  for i in range(1, 11):
    fila = []
    for c in range(1, 11):
      resultado = 0
      for c in range(c):
        resultado = resultado + i
      fila.append(resultado)
    matriz.append(fila)


def imprimir_tabla(matriz):
  print(" ", end="")
  for c in range(1, 11):
    print(f"{c:2}", end=" ")
  print()

  for i in range(1, 11):
    print(f"{i:2} ", end="")
    for c in range(1, 11):
      print(f"{matriz[i-1][c-1]:2}", end=" ")
    print()


def consultar_producto(numero1, numero2, matriz):
  return matriz[numero1 - 1][numero2 - 1]

llenar_matriz(matriz)
imprimir_tabla(matriz)

print("")
# Ciclo while para rechazar 0, 12 o errores y volver a preguntar
while True:
  try:
    numero1 = int(input("Ingrese el primer numero del 1 a 10: "))
    numero2 = int(input("Ingrese el segundo numero del 1 a 10: "))
    if 1 <= numero1 <= 10 and 1 <= numero2 <= 10:
      resultado = consultar_producto(numero1, numero2, matriz)
      print(f"El resultado es: {resultado} ")
      break
      # Finaliza correctamente cuando los datos son válidos
    else:
      print("Numero no valido: Los numeros deben estar entre 1 y 10.")
  except ValueError:
    print( "Error: Por favor, ingrese solo un número entero válido para cada" " factor." )
```
