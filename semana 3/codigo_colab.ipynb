```python
# RELIABLE FINANCE SOFTWARE

print("===================================")
print("     RELIABLE FINANCE SOFTWARE")
print("===================================")

# Registrar ingresos
total_ingresos = float(input("Ingrese el total de ingresos familiares: $"))

while total_ingresos <= 0:
    print("Cantidad no valida. Debe ser mayor que cero.")
    total_ingresos = float(input("Ingrese nuevamente sus ingresos: $"))

# Registrar gastos
total_gastos = float(input("Ingrese el total de gastos mensuales: $"))

while total_gastos <= 0:
    print("Cantidad no valida. Debe ser mayor que cero.")
    total_gastos = float(input("Ingrese nuevamente sus gastos: $"))

# Calcular saldo
saldo = total_ingresos - total_gastos

# Calcular porcentaje disponible
porcentaje = (saldo / total_ingresos) * 100

# Mostrar resultados
print("\n===================================")
print("       RESUMEN FINANCIERO")
print("===================================")

print("Total de ingresos: $", total_ingresos)
print("Total de gastos: $", total_gastos)
print("Saldo disponible: $", saldo)
print("Porcentaje disponible:", porcentaje, "%")

print("===================================")
print("Gracias por utilizar Reliable Finance Software")
print("Programa finalizado.")
```
